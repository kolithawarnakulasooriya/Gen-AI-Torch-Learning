```python
from tokenizers import Tokenizer, models,trainers

special_tokens=[
    "<unk>",
    "<startoftext>",
    "<endoftext>",
    "<sep>",
    "<pad>",
    "<cls>",
    ]

tokenizer = Tokenizer(models.BPE())
trainer = trainers.BpeTrainer(
    vocab_size=1024,
    special_tokens=special_tokens
)
tokenizer.train([ "data.txt"], trainer=trainer)
```

```python
tokenizer.save("tokenizer")
```
